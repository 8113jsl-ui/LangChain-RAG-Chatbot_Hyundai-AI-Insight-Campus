```python
## 문항 6. 첫 API 호출 구현 ##


## .env 파일 불러오기
from dotenv import load_dotenv
import os

load_dotenv()

print("API 키 로드 여부:", bool(os.getenv("OPENAI_API_KEY")))




## client 생성
from openai import OpenAI

client = OpenAI()
print("✅ OpenAI 클라이언트 준비 완료")





response = client.chat.completions.create(  # client.chat.completions.create (사용자 채팅에 대해 응답 생성) 메서드
    model="gpt-4o-mini",       # 사용할 모델 지정             
    messages=[                                  
        {"role": "user", "content": "RAG가 무엇인가요?"}   # 메시지 역할 user의 request를 messages 리스트에 할당
    ]
)


print(response.choices[0].message.content)  # 응답 메시지 출력

'''
RAG는 Retrieval-Augmented Generation의 약어로, 정보 검색과 생성 모델을 결합한 인공지능 기술을 의미합니다. 이 방법론은 주로 자연어 처리(NLP)에서 사용되며, 특정 질문에 대한 답변을 생성하기 위해 외부 데이터베이스나 문서에서 관련 정보를 검색하는 방식입니다.

RAG의 작동 방식은 일반적으로 다음과 같습니다:

1. **정보 검색**: 질문이나 요청에 대한 관련 정보를 데이터베이스나 문서에서 검색합니다. 이 과정에서 검색 엔진이나 고급 텍스트 검색 알고리즘을 사용할 수 있습니다.

2. **정보 생성**: 검색된 정보를 바탕으로 답변을 생성합니다. 이 단계에서는 자연어 생성(NLG) 모델이 사용되어, 사용자가 이해할 수 있는 형식으로 정보를 제공합니다.

RAG는 대규모 언어 모델의 생성 능력과 유연성을 활용하면서도, 외부 정보에 접근하여 더 정확하고 최신의 정보를 제공할 수 있는 장점이 있습니다. 이러한 접근 방식은 사실 기반의 질문 응답 시스템, 대화형 AI, 그리고 다양한 응용 프로그램에서 유용하게 활용될 수 있습니다.
'''
```

```python
## 문항 7. 대화 맥락 유지 ##

def ask_llm(messages, temperature=0.3, max_tokens=500):
    """메시지 리스트를 받아서 LLM의 답변 텍스트를 돌려준다."""
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
        temperature=temperature,
        max_tokens=max_tokens,
    )
    return response.choices[0].message.content


messages = [
    {"role": "system", "content": "당신은 담담한 어조로 답변합니다."},   # system은 챗봇의 톤을 지정합니다.
    {"role": "user", "content": "제 이름은 이재서입니다."},             # user는 사용자의 요청 메시지 처리에 사용됩니다.
    {"role": "assistant", "content": "안녕하세요, 재서님. 오늘 하루 잘 지내셨나요?"},   # assistant는 LLM이 생성한 답변 메시지 처리에 사용됩니다.
    {"role": "user", "content": "제 이름이 뭐였죠?"}  # ← LLM은 위 대화를 보고 답변
]

print(ask_llm(messages))
'답변 : 당신의 이름은 이재서입니다.'

```

```python
## 문항 8. 고객 문의 분류기 완성 ##

# 분류기 함수
def classify_inquiry(text: str) -> str:
    """고객 문의를 4개 카테고리 중 하나로 분류"""
    messages = [
        {"role": "system", "content": "고객 문의를 다음 4개 카테고리 중 하나로 분류해야 합니다: 결제/배송/환불/기타"},
        {"role": "user", "content": text},
    ]
    return ask_llm(messages, temperature=0, max_tokens=10)

# 실행할 테스트 예제들
test_cases = [
    "어제 주문했는데 카드 결제가 두 번 되었어요.",       # 예측: ____
    "택배가 일주일째 안 와요.",                          # 예측: ____
    "사이즈가 안 맞아서 반품하고 싶어요.",                # 예측: ____
    "영업시간이 어떻게 되나요?", 
    "결제했는데 취소하고 다른 카드로 다시 결제하고 싶어요."
]

# 실행
for text in test_cases:
    category = classify_inquiry(text)
    print(f"[{category:>4}] {text}")
    
    
'''
[  결제] 어제 주문했는데 카드 결제가 두 번 되었어요.
[  배송] 택배가 일주일째 안 와요.
[  환불] 사이즈가 안 맞아서 반품하고 싶어요.
[  기타] 영업시간이 어떻게 되나요?
[결제/배송/환불/기타] 결제했는데 취소하고 다른 카드로 다시 결제하고 싶어요.
'''

# 애매한 분류 : "결제했는데 취소하고 다른 카드로 다시 결제하고 싶어요."
# [결제/환불]로 분류되어야 하는데, [결제/배송/환불/기타]로 엄밀히 필터링되지 않았습니다.
```

```python
## 문항 9. temperature 비교 실험 ##

prompt = "AI 스타트업 이름 3개를 지어주세요."

for temp in [0, 1.2]:
    answer = ask_llm(
        [{"role": "user", "content": prompt}],
        temperature=temp,
    )
    print(f"🌡 temperature={temp}\n→ {answer}\n")
    


'''
🌡 temperature=0
→ 물론입니다! 다음은 AI 스타트업을 위한 이름 세 가지입니다:

1. **인사이트AI (InsightAI)** - 데이터 분석과 인사이트 제공에 중점을 둔 스타트업에 적합한 이름입니다.
  
2. **넥스트젠AI (NextGenAI)** - 차세대 기술과 혁신을 강조하는 스타트업에 어울리는 이름입니다.

3. **스마트코어 (SmartCore)** - AI의 핵심 기술을 개발하거나 제공하는 스타트업에 적합한 이름입니다.

이 이름들이 도움이 되길 바랍니다!

🌡 temperature=1.2
→ 물론입니다! 다음은 AI 스타트업을 위한 이름 제안 3가지입니다:

1. **IntelliVerse**: 지능과 우주(verse)의 결합으로, 다양한 AI 솔루션을 탐구하고 혁신하는 이미지를 줍니다.
  
2. **CogniTech**: "Cogni"는 인지(cognition)의 약어로, 지능적인 기술을 개발하고 제공하는 스타트업을 표현합니다. 

3. **AetherAI**: "Aether"는 고대 그리스에서 하늘이나 에너지를 상징하던 개념으로, 인공지능의 미래적인 특성을 강조하는 이름입니다.

이름이 마음에 드시길 바랍니다! 추가적인 도움이 필요하시면 언제든지 말씀해 주세요.
'''
```

**temperature = 0 vs temperature = 1.2**

- temperature = 0은 회사명에 'AI'나 '스마트'가 포함되는 일관되나 다소 뻔한 결과가 나왔습니다.
- temperature = 1.2는 Intelliverse, Cogni, Aether와 같이 합성어, 축약어, 그리스어 등 다양한 시도를 했으며 'AI'라는 단어를 직접 사용하지는 않았습니다.

```python
## 문항 10. 호출 비용 계산 ##

from cProfile import label


PRICE_INPUT = 0.15   # 입력 1M 토큰당 USD
PRICE_OUTPUT = 0.60  # 출력 1M 토큰당 USD
USD_TO_KRW = 1400    # 환율

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": "말티즈가 참지 않는 이유에 대해 알려줘!"}]
)
def cost_korean_won(usage):
    usage = response.usage 
    input_cost = usage.prompt_tokens / 1_000_000 * PRICE_INPUT
    output_cost = usage.completion_tokens / 1_000_000 * PRICE_OUTPUT
    return (input_cost + output_cost) * USD_TO_KRW

u = response.usage
print(f"[{label}] 입력 {u.prompt_tokens} + 출력 {u.completion_tokens} = "
f"총 {u.total_tokens} 토큰, 비용 약 {cost_korean_won(u):.4f}원")
```

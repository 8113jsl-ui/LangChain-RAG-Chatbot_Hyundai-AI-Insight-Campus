```python
# langchain 패키지 가져오기
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from datetime import datetime

print("✅ import 완료")
```

```python
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.3, max_tokens=200)

prompt = ChatPromptTemplate.from_messages([
    ("system", "당신은 {role}입니다."),
    ("human", "{question}")
])

chain = prompt | llm
```

```python
# 페르소나 사전: 이름 → system 메시지
PERSONAS = {
    "분석가": "당신은 10년 차 데이터 분석가입니다. 수치와 근거를 들어 간결하게 답변하세요.",
    "마케터": "당신은 디지털 마케팅 전문가입니다. 비즈니스 관점과 사례를 들어 답변하세요.",
    "개발자": "당신은 시니어 백엔드 개발자입니다. 코드 예시와 함께 실용적으로 답변하세요.",
    "기획자": "당신은 IT 서비스 기획자입니다. 사용자 가치와 우선순위 관점에서 답변하세요.",
    "디자이너": "당신은 UX 디자이너입니다. 사용자 경험과 시각적 비유로 쉽게 답변하세요.",
}

print(f"✅ {len(PERSONAS)}개 페르소나 등록 완료: {list(PERSONAS.keys())}")

# llm은 미리 생성한 LangChain 채팅 모델
PERSONA_LLMS = {
    "분석가": llm.bind(temperature=0),
    "마케터": llm.bind(temperature=0.6),
    "개발자": llm.bind(temperature=0.3),
    "기획자": llm.bind(temperature=0.7),
    "디자이너": llm.bind(temperature=0.9),
}

print(f"✅ {len(PERSONA_LLMS)}개 temperature 등록 완료: {list(PERSONA_LLMS.keys())}")
```

```python
PERSONA_INTROS = {
    "분석가": "수치와 근거 기반 답변",
    "마케터": "비즈니스 관점과 마케팅 사례 중심 답변",
    "개발자": "코드 예시를 활용한 실용적인 답변",
    "기획자": "사용자 가치와 우선순위 중심 답변",
    "디자이너": "사용자 경험과 시각적 비유로 쉬운 답변",
}
```

### 실습과제 1번, 실습과제 2번

```python
log = []

while True:
    persona = input("페르소나 선택 (/list: 목록, 종료: 종료): ").strip()

    if persona == "종료":
        break

    if persona == "/list":
        for name in PERSONAS:
            print(f"- {name}: {PERSONA_INTROS.get(name, '소개 미등록')}")
        continue
    
    if persona == "/help":
        print("사용법:")
        print("- 페르소나 선택 후 질문 입력")
        print("- /list: 페르소나 목록 확인")
        print("- 종료: 프로그램 종료")
        print("\n")
        print("명령어 목록:")
        print("- /quit, /exit: 프로그램 종료")
        print("- /personas: 페르소나 목록 확인")
        print("- /list: 페르소나 목록 확인")
        print("- /help: 도움말 보기")

        print("- /: 도움말 보기")
        print("- /help: 도움말 보기")
        print("- /help: 도움말 보기")
                                        
        continue

    if persona not in PERSONAS:
        print("목록에 있는 페르소나를 선택하세요. (/list: 목록)")
        continue

    user_input = input("질문: ")

    chain = prompt | PERSONA_LLMS[persona] | StrOutputParser()

    answer = chain.invoke({
        "role": PERSONAS[persona],
        "question": user_input,
    })

    print(answer)

    log.append(
        f"[{datetime.now().strftime('%H:%M:%S')}] "
        f"[{persona}] {user_input} → {answer}"
    )

# 종료 시 로그 저장
with open("chat_log.txt", "w", encoding="utf-8") as f:
    f.write("\n".join(log))
```

### 실습과제 4번

```python
def handle_command(user_input: str, current_persona: str):
    parts = user_input.split(maxsplit=1)   # "/change 마케터" → ["/change", "마케터"]
    cmd = parts[0]

    if cmd in ("/quit", "/exit"):
        print("👋 종료합니다.")
        return current_persona, True       # 종료 플래그 True

    if cmd == "/persona":
        print(f"🎭 현재 페르소나: {current_persona}")

    elif cmd == "/help":
        print("📖 명령어: /quit, /exit, /change <이름>, /persona, /list, /help")

    elif cmd == "/change":
        new_name = parts[1] if len(parts) > 1 else ""
        if new_name in PERSONAS:
            print(f"✅ 페르소나가 '{new_name}'(으)로 변경되었습니다.")
            return new_name, False         # 페르소나만 변경
        else:
            print(f"❌ '{new_name}'은 등록된 페르소나가 아닙니다.")
            print(f"   사용 가능: {', '.join(PERSONAS.keys())}")

    elif cmd == "/list":
        print("\n사용 가능한 페르소나:")
        for name in PERSONAS:
            print(f"- {name}: {PERSONA_INTROS.get(name, '소개 미등록')}")
        print()

    else:
        print(f"❓ 알 수 없는 명령어: {cmd} (/help 입력)")

    return current_persona, False
```
